```python
from courtside_dynamics.colab_setup import setup_colab
setup_colab()

from courtside_dynamics.envs import TennisWallEnv
from courtside_dynamics.training import TrainConfig, train

cfg = TrainConfig(
    env_fn=lambda: TennisWallEnv(render_mode="rgb_array", min_force=100.0),
    algo="SAC",
    total_timesteps=2_000_000,
    log_dir="./logs/TennisWall_SAC",
    name_prefix="tennis_wall_sac",
    eval_freq=25_000,
    csv_header=[
        "phase", "rally_count",
        "paddle_hit_count", "wall_hit_count",
        "paddle_touch", "wall_touch",
        "reward", "total_reward", "done",
    ],
    info_row_fn=lambda info, r, tr, d: [
        info["phase"], info["rally_count"],
        info["paddle_hit_count"], info["wall_hit_count"],
        info["paddle_touch"], info["wall_touch"],
        r, tr, d,
    ],
)
model = train(cfg)
```
